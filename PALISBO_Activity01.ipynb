```python
import numpy as np

def my_dist_xyz(x,y,z):

  def my_dist(x,y):
    out = np.sqrt((x[0]-y[0])**2+(x[1]-y[1])**2)
    return out

  d0 = my_dist(x,y)
  d1 = my_dist(x,z)
  d2 = my_dist(y,z)

  return [d0, d1, d2]

d = my_dist_xyz((0,0), (0,1),(1,1))
print(d)


```

```python
def my_tip_calc(bill, party):
  if party < 6:
    tips = bill * 0.15
  elif party < 8:
    tips = bill * 0.18
  elif party <11:
    tips = bill * 0.20
  elif party > 11:
    tips = bill * 0.25
  return tips

t = my_tip_calc(109.29,3)
print(t)


z = my_tip_calc(109.29,7)
print(z)

x = my_tip_calc(109.29,9)
print(x)

y = my_tip_calc(109.29,12)
print(y)
```
