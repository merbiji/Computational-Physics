```python
import numpy as np

def my_dist_xyz(x,y,z):

  def my_dist(x,y):
    out = np.sqrt((x[0]-y[0])**2+(x[1]-y[1])**2)
    return out

  d0 = my_dist(x,y)
  d1 = my_dist(x,z)
  d2 = my_dist(y,z)

  return [d0, d1, d2]

d = my_dist_xyz((0,0), (0,1),(1,1))
print(d)


```

```python

def my_thermo_stat(temp, desired_temp):
    if temp < desired_temp - 5:
        temp_status = 'Heat'
    elif    temp > desired_temp + 5:
        temp_status = "AC"
    else:
        temp_status = "OFF"

    return temp_status

status = my_thermo_stat(75,65)

print(status)
```
