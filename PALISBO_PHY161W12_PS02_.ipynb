```python
import matplotlib.pyplot as plt
import math

def compute_sin(x, tol =1e-7):
  term = x
  total_sum = x
  n = 2

  exact = math.sin(x)
  denom = abs(exact) if exact != 0 else 1.0

  errors = [max(abs(exact - total_sum) / denom, 1e-16)]

  print(f"\n------Sin approximation for x = {x}------")
  print(f"n = 1, \t sum = {total_sum:.7f},\t term = {term:.7f}")


  while True:
    term = -term * x **2 / ((2*n - 1)*(2*n - 2))
    total_sum += term

    rel_err = abs(exact - total_sum) / denom
    errors.append(max(rel_err, 1e-16))


    print(f"n = {n},\t sum = {total_sum:.7f},\t term = {term:.7f}")

    if abs(term/total_sum) <= tol:
      break
    n += 1

  return errors


X = [1, 3, 5, 10, 20, 50, 100]
plt.figure(figsize=(10, 6))

for x in X:
  err_list = compute_sin(x)
  plt.plot(range(1, len(err_list) + 1), err_list, marker='o', label=f'x = {x}')

plt.yscale('log')
plt.xlabel("Number of Terms(N)")
plt.ylabel("Errors")
plt.grid(True)
plt.legend
plt.show
```
