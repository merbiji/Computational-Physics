```python
import numpy as np

class derivative_approx:
    
    @staticmethod
    def f1(x):
        return np.tan(x / 4)
    
    @staticmethod
    def f2(x):
        return 1/x * np.sin(0.4*np.sqrt(x))
    
    @staticmethod
    def f3(x):
        return np.exp(0.5*x) + x
    
    @staticmethod
    def df1(x):
        return 0.25 / np.cos(x / 4) ** 2

    @staticmethod
    def df2(x):
        return -np.sin(0.4 * np.sqrt(x)) / x**2 + 0.2 * np.cos(0.4 * np.sqrt(x)) / x**1.5

    @staticmethod
    def df3(x):
        return 0.5 * np.exp(0.5 * x) + 1
    @staticmethod
    def forward_difference(f,x,h):
        return (f(x+h) - f(x) )/ h
    
    @staticmethod
    def central_difference(f,x,h):
        return (f(x + h) - f(x - h)) / (2 * h)

    def compare(self):
        cases = [
            ("(i)   tan(x/4)",              self.f1, self.df1, 2.0, 0.4),
            ("(ii)  (1/x) sin(0.4 sqrt(x))", self.f2, self.df2, 1.5, 0.2),
            ("(iii) e^(0.5x) + x",          self.f3, self.df3, 2.0, 0.2),
        ]
        for label, f, df, x, h in cases:
            exact = df(x)
            fd = self.forward_difference(f, x, h)
            cd = self.central_difference(f, x, h)
            print(f"{label}   x = {x}, h = {h}")
            print(f"  Exact   : {exact:.8f}")
            print(f"  Forward : {fd:.8f}   |error| = {abs(fd - exact):.3e}")
            print(f"  Central : {cd:.8f}   |error| = {abs(cd - exact):.3e}\n")


if __name__ == "__main__":
    derivative_approx().compare()

```

```python
import math

# (label, f, exact f', x, h)
cases = [
    ("(i)   f(x) = tan(x/4)",
     lambda x: math.tan(x / 4),
     lambda x: 0.25 / math.cos(x / 4) ** 2,
     2.0, 0.4),
    ("(ii)  f(x) = (1/x) sin(0.4 sqrt(x))",
     lambda x: math.sin(0.4 * math.sqrt(x)) / x,
     lambda x: -math.sin(0.4 * math.sqrt(x)) / x**2
               + 0.2 * math.cos(0.4 * math.sqrt(x)) / x**1.5,
     1.5, 0.2),
    ("(iii) f(x) = e^(0.5x) + x",
     lambda x: math.exp(0.5 * x) + x,
     lambda x: 0.5 * math.exp(0.5 * x) + 1,
     2.0, 0.2),
]

def forward(f, x, h):
    return (f(x + h) - f(x)) / h

def central(f, x, h):
    return (f(x + h) - f(x - h)) / (2 * h)

for label, f, df, x, h in cases:
    exact = df(x)
    fd, cd = forward(f, x, h), central(f, x, h)
    print(label, f"   x = {x}, h = {h}")
    print(f"  Exact derivative     : {exact:.8f}")
    print(f"  Forward difference   : {fd:.8f}   |error| = {abs(fd - exact):.3e}")
    print(f"  Central difference   : {cd:.8f}   |error| = {abs(cd - exact):.3e}")
    better = "central" if abs(cd - exact) < abs(fd - exact) else "forward"
    print(f"  More accurate        : {better}\n")
```
