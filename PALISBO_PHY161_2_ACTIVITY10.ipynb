```python
import numpy as np

A_1 = np.array([[2,-6,-1],[-3,-1,7],[-8,1,-2]], dtype =float)
k_1 = np.array([38,-34,-20], dtype=float)

A_2 = np.array([[2,1,-1],[5,2,2],[3,1,1]], dtype =float)
k_2 = np.array([1,-4,-5], dtype=float)

A_3 = np.array([[2,-4,5],[1,1.5,-7],[5,3,-8]], dtype =float)
k_3 = np.array([36,7,-31], dtype=float)

systems = [
    (A_1, k_1,),
    (A_2, k_2),
    (A_3, k_3)
]

def gauss_elimination(A,k):
  A = A.copy()
  k = k.copy()
  n = len(A)

  A = np.column_stack((A,k))

  # Forward Elimination
  for i in range(n):
    pivot = A[i][i]

    if pivot == 0:
      for r in range(i + 1, n):
        if A[r][i] != 0:
          A[i], A[r] = A[r].copy(),A[i].copy()
          pivot = A[i][i]
          break

    if pivot == 0:
      raise ValueError("No unique solution exists.")

    # Normalize pivot row
    for j in range(i, n+1):
      A[i][j] = A[i][j] / pivot

    for r in range(i + 1, n):
      factor = A[r][i]

      for j in range(i, n+1):
        A[r][j] = A[r][j] - factor * A[i][j]

  print("Final Augmented Matrix:")
  print(A)


  x = [0] * n

  for i in range(n -1, -1, -1):
    x[i] = A[i][n]

    for j in range(i + 1, n):
      x[i] -= A[i][j] * x[j]

  return x

def gauss_jordan(A, k):
    A = A.copy()
    k = k.copy()
    n = len(A)

    A = np.column_stack((A, k))

    # Forward and Backward Elimination
    for i in range(n):

        pivot = A[i][i]

        # Pivoting
        if pivot == 0:
            for r in range(i + 1, n):
                if A[r][i] != 0:
                    A[i], A[r] = A[r].copy(), A[i].copy()
                    pivot = A[i][i]
                    break

        if pivot == 0:
            raise ValueError("No unique solution exists.")

        # Normalize pivot row
        for j in range(i, n+1):
            A[i][j] = A[i][j] / pivot

        # Eliminate entries above and below pivot
        for r in range(n):
            if r != i:
                factor = A[r][i]

                for j in range(i, n+1):
                    A[r][j] = A[r][j] - factor * A[i][j]

    print("Final Augmented Matrix (Gauss-Jordan):")
    print(A)

    # Extract solutions
    x = A[:, n]

    return x

for index, (A, k) in enumerate(systems, start=1):

    print(f"\nSYSTEM {index}")
    print("="*60)

    solution_gauss = gauss_elimination(A, k)
    solution_gauss_jordan = gauss_jordan(A, k)

    print("\nGaussian Elimination Solution:")
    for i in range(len(solution_gauss)):
        print(f"x{i+1} = {solution_gauss[i]}")

    check = np.linalg.solve(A, k)


    print("\nGaussian-Jordan Solution:")
    for i in range(len(solution_gauss_jordan)):
        print(f"x{i+1} = {solution_gauss_jordan[i]}")

    print("\nNumPy Solution:")
    for i in range(len(check)):
        print(f"x{i+1} = {check[i]}")
```
