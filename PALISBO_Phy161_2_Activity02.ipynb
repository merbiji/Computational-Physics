```python
import math

def compute_sin(x , N = 20):
  term = x
  sin_sum = x
  sin_exact = math.sin(x)

  print(f"\t\n---------Sin approximation for x = {x:.2f}---------")
  print("-"*50)
  print(f"iteration:   1, sum: {sin_sum:.6f}, abs diff:{abs(sin_sum - sin_exact):.6e}")


  for n in range(2, N + 1):
    term = -term * x * x / ((2 * n - 2) * (2 * n - 1))
    sin_sum += term
    print(f"iteration: {n:3}, sum: {sin_sum:.6f}, abs diff:{abs(sin_sum - sin_exact):.6e}")

  return sin_sum, n


X = [0.1, 1.0, 2.0]

for x in X:
  compute_sin(x)
```

```python
import math

def compute_cos(x , N = 20):
  y = 2 * x
  term = 1
  cos_sum = 1
  cos_exact = math.cos(y)

  print(f"\t\n---------cosine approximation for x = {x:.2f}---------")
  print("-"*50)
  print(f"iteration:   1, sum: {cos_sum:.6f}, abs diff:{abs(cos_sum - cos_exact):.6e}")


  for n in range(2, N + 1):
    term = -term * y * y / ((2 * n - 2) * (2 * n - 1))
    cos_sum += term
    print(f"iteration: {n:3}, sum: {cos_sum:.6f}, abs diff:{abs(cos_sum - cos_exact):.6e}")

  return cos_sum, n


X = [0.1, 1.0, 2.0]

for x in X:
  compute_cos(x)
```

```python
import math

def compute_x2sin(x , N = 20):
  term = x
  sin_sum = x
  sin_exact = math.sin(x) * (x**2)
  total_sum = x*x*sin_sum

  print(f"\t\n---------x^2sin(x) approximation for x = {x:.2f}---------")
  print("-"*50)
  print(f"iteration:   1, sum: {total_sum:.6f}, abs diff:{abs(total_sum - sin_exact):.6e}")


  for n in range(2, N + 1):
    term = -term * x * x / ((2 * n - 2) * (2 * n - 1))
    sin_sum += term
    total_sum = x*x*sin_sum
    print(f"iteration: {n:3}, sum: {total_sum:.6f}, abs diff:{abs(total_sum - sin_exact):.6e}")

  return total_sum, n


X = [0.1, 1.0, 2.0]

for x in X:
  compute_x2sin(x)
```
