```python
import numpy as np
import math
import matplotlib.pyplot as plt

Xmax = 40
Xmin = 0.25
step = 0.1
order = 10
start = 50

def down(x, n, m):
    j = np.zeros((start + 2), float)
    j[m+1] = j[m] = 1.0

    if x == 0.0:
        return 1.0 if n == 0 else 0.0

    for k in range(m, 0, -1):
        j[k-1] = ((2.0 * k + 1.0) / x) * j[k] - j[k+1]

    scale = (math.sin(x) / x) / j[0]
    return scale * j[n]

def up(x, n, max_order):
    j = np.zeros((max_order + 1), float)

    if x == 0.0:
        return 1.0 if n == 0 else 0.0

    j[0] = math.sin(x) / x
    j[1] = (math.sin(x) / x**2) - (math.cos(x) / x)

    for k in range(1, max_order):
        j[k+1] = ((2.0 * k + 1.0) / x) * j[k] - j[k-1]

    return j[n]

x_vals = np.arange(Xmin, Xmax, step)
y_red = [down(x, order, start) for x in x_vals]
y_blue = [down(x, 1, start) for x in x_vals]
y_green = [up(x, order, start) for x in x_vals]
y_orange = [up(x, 1, start) for x in x_vals]
plt.figure(figsize=(9, 5))
plt.plot(x_vals, y_red, color='red', label=f"j_{order}(x) (down)")
plt.plot(x_vals, y_blue, color='blue', label="j_1(x) (down)")
plt.plot(x_vals, y_green, color='green', label=f"j_{order}(x) (up)")
plt.plot(x_vals, y_orange, color='orange', label="j_1(x) (up)")
plt.ylim(-0.2, 0.5)
plt.xlim(Xmin, Xmax)
plt.title("Spherical Bessel, L = 1, 10 (downward and upward recurrence)")
plt.xlabel("x")
plt.ylabel("j(x)")
plt.legend()
plt.grid(True, linestyle=':', alpha=0.6)
plt.show()
```

```python
import numpy as np
import math
import matplotlib.pyplot as plt


x_values = [0.1, 1.0, 10.0, 20.0]

def down(x, n, m):
    j = np.zeros((start + 2), float)
    j[m+1] = j[m] = 1.0

    if x == 0.0:
        return 1.0 if n == 0 else 0.0

    for k in range(m, 0, -1):
        j[k-1] = ((2.0 * k + 1.0) / x) * j[k] - j[k+1]

    scale = (math.sin(x) / x) / j[0]
    return scale * j[n]

def up(x, n, max_order):
    j = np.zeros((max_order + 1), float)

    if x == 0.0:
        return 1.0 if n == 0 else 0.0

    j[0] = math.sin(x) / x
    j[1] = (math.sin(x) / x**2) - (math.cos(x) / x)

    for k in range(1, max_order):
        j[k+1] = ((2.0 * k + 1.0) / x) * j[k] - j[k-1]

    return j[n]

for test_x in x_values:
    print(f"\n{'='*65}")
    print(f"Results for x = {test_x}")
    print(f"{'l':<4} | {'j_l (up)':<15} | {'j_l (down)':<15} | {'Relative Difference':<20}")
    print(f"{'-'*65}")
    for l in range(25):
        up_val = up(test_x, l, max_order=24)
        down_val = down(test_x, l, start)
        denom = abs(up_val) + abs(down_val)
        rel_diff = abs(up_val - down_val) / denom if denom != 0 else 0.0
        print(f"{l:<4} | {up_val:<15.4e} | {down_val:<15.4e} | {rel_diff:<20.4e}")

```
