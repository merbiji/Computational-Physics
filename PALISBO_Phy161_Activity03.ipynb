```python
from matplotlib import lines
import matplotlib.pyplot as plt
import numpy as np
import math
def summation_S1(N):

  total_sum = 0

  for n in range(1, 2*N+1):

    sign = 1.0 if n%2 == 0 else -1.0
    term = n / (n + 1)
    total_sum += sign*term

  return total_sum

def summation_S2(N):
  sum1 = 0
  sum2 = 0
  for n in range(1, N+1):
    term1 = (2*n - 1) / (2*n)
    term2 = 2*n / (2*n + 1)
    sum1 += term1
    sum2 += term2

  return -sum1 + sum2

def summation_S3(N):
  total_sum = 0
  for n in range(1, N+1):
    term = 1 / ((2*n)*((2*n) + 1))
    total_sum += term
  return total_sum

N_values = [1,5,10,20, 50,100,500,1000,5000,10000,50000,100000,500000,1000000]
rel_err_S1 = []
rel_err_S2 = []

for N in N_values:
  s1 = summation_S1(N)
  s2 = summation_S2(N)
  s3 = summation_S3(N)

  err1 = abs((s1-s3)/ s3)
  err2 = abs((s2-s3)/ s3)

  rel_err_S1.append(max(err1, 1e-16))
  rel_err_S2.append(max(err2, 1e-16))


plt.figure(figsize=(9, 6))
plt.loglog(N_values, rel_err_S1, marker='o', label='Relative error in S1')
plt.loglog(N_values, rel_err_S2, marker='s', label='Relative error in S2')
plt.xscale('log')
plt.yscale('log')
plt.xlabel('Number of Terms (N)')
plt.ylabel('Relative Error')
plt.title('Relative Error vs N')
plt.legend()
plt.grid(True, linestyle = '--')
plt.show()
```
