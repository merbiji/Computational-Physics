1.a & b

```python
import math

def quadratic_x(a,b,c):
  dsc = b**2 - 4*a*c
  if dsc < 0:
    print("No real solutions")
    return
  x1 = (-b + math.sqrt(dsc)) / (2*a)
  x2 = (-b - math.sqrt(dsc)) / (2*a)
  return x1, x2

def quadratic_xprime(a,b,c):
  dsc = b**2 - 4*a*c
  if dsc < 0:
    print("No real solutions")
    return

  x1 = -2*c / (b - math.sqrt(dsc))
  x2 = -2*c / (b + math.sqrt(dsc))
  return x1, x2

a = 1
b = 1
print("FORMULA 1")
print("-" * 70)
for n in range(1, 16):
  c = 10**(-n)
  x1, x2 = quadratic_x(a,b,c)
  print(f'c:{c:.2e}\t   x1: {x1:.16e}   \t x2: {x2:.16e}')
print("\nFORMULA 2")
print("-" * 70)
for n in range(1, 16):
  c = 10**(-n)
  x1, x2 = quadratic_xprime(a,b,c)
  print(f'c:{c:.2e}\t   x1: {x1:.16e}   \t x2: {x2:.16e}')
```

1. c

```python
import math
import matplotlib.pyplot as plt


def quadratic_x(a,b,c):
  dsc = b**2 - 4*a*c
  if dsc < 0:
    print("No real solutions")
    return
  if a == 0:
    print("Not a quadratic equation")
    return
  sgn_b = 1.0 if b >= 0 else -1.0
  q = -0.5 * (b + sgn_b * math.sqrt(dsc))


  x1 = q / a
  x2 = c / q if q != 0 else 0.0



  return x1, x2

def quadratic_xprime(a,b,c):
  dsc = b**2 - 4*a*c
  if dsc < 0:
    print("No real solutions")
    return
  if a == 0:
    print("Not a quadratic equation")
    return
  sgn_b = 1.0 if b >= 0 else -1.0
  q = -0.5 * (b + sgn_b * math.sqrt(dsc))


  x1 = c /q if q != 0 else 0.0
  x2 = q/ a



  return x1, x2

a = 1.0
b = 1.0
err1 = []

print(f"{'c'}      \t|\t {'Form 1 (x1)':<25} |\t {'Form 2 (Exact x1)':<25} |\t{'Relative Error':<20}")
print("-" * 100)
for n in range(1, 16):
  c = 10**(-n)

  x1, x2 = quadratic_x(a,b,c)
  x1prime, x2prime = quadratic_xprime(a,b,c)
  denom = abs(x1prime) if x1prime != 0 else 1e-16
  rel_err = abs(x1 - x1prime) / denom

  err1.append(rel_err)


  print(f"{c} \t|\t {x1:<25.16e} |\t {x1prime:<25.16e} |\t{rel_err:<20.10e}")

plt.figure(figsize=(9, 6))
plt.loglog(err1, marker='o')
plt.xscale('log')
plt.grid(True)
plt.yscale('log')
plt.xlabel('c values')
plt.ylabel('Relative Error')
plt.title('Relative Error vs c')
plt.show()
```

```python
from matplotlib import lines
import matplotlib.pyplot as plt
import numpy as np
import math
def summation_S1(N):

  total_sum = 0

  for n in range(1, 2*N+1):

    sign = 1.0 if n%2 == 0 else -1.0
    term = n / (n + 1)
    total_sum += sign*term

  return total_sum

def summation_S2(N):
  sum1 = 0
  sum2 = 0
  for n in range(1, N+1):
    term1 = (2*n - 1) / (2*n)
    term2 = 2*n / (2*n + 1)
    sum1 += term1
    sum2 += term2

  return -sum1 + sum2

def summation_S3(N):
  total_sum = 0
  for n in range(1, N+1):
    term = 1 / ((2*n)*((2*n) + 1))
    total_sum += term
  return total_sum

N_values = [1,5,10,20, 50,100,500,1000,5000,10000,50000,100000,500000,1000000]
rel_err_S1 = []
rel_err_S2 = []

for N in N_values:
  s1 = summation_S1(N)
  s2 = summation_S2(N)
  s3 = summation_S3(N)

  err1 = abs((s1-s3)/ s3)
  err2 = abs((s2-s3)/ s3)

  rel_err_S1.append(max(err1, 1e-16))
  rel_err_S2.append(max(err2, 1e-16))


plt.figure(figsize=(9, 6))
plt.loglog(N_values, rel_err_S1, marker='o', label='Relative error in S1')
plt.loglog(N_values, rel_err_S2, marker='s', label='Relative error in S2')
plt.xscale('log')
plt.yscale('log')
plt.xlabel('Number of Terms (N)')
plt.ylabel('Relative Error')
plt.title('Relative Error vs N')
plt.legend()
plt.grid(True, linestyle = '--')
plt.show()
```
