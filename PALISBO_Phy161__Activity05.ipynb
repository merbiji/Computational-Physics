For Phy161.2 MW61-2

(1) Bisection: f(x)=x3−7x+1=0, interval: [0,1], tolerance: 10−6

(2) Newton-Raphnson: f(x)=cosx−x, Initial guess: x0=0.5, tolerance: 10−6

(3) Secant: f(x)=x2−5=0, Initial guess: x0=2,x1=3, tolerance: 10−6

```python
def function(x):
    return x**3 - 7*x + 1

def bisection(x_u, x_l, tol=1e-6, max_iter=100):
    f_l = function(x_l)
    if function(x_u) * f_l >= 0.0:
        print("No root bracketed")
        return None

    print(f"{'Iter':<5}| {'x_l':>10} | {'x_u':>10} | {'x_m':>10} | {'f(x_m)':>12} | {'f(xl)*f(xm)':>12} | {'check':>5}")

    for n in range(1, max_iter + 1):
        x_m = (x_u + x_l) / 2.0
        f_m = function(x_m)
        product = f_l * f_m

        if product < 0.0:
            sign = "< 0"
        elif product > 0.0:
            sign = "> 0"
        else:
            sign = "= 0"

        print(f"{n:<5}| {x_l:>10.6f} | {x_u:>10.6f} | {x_m:>10.6f} | {f_m:>12.3e} | {product:>12.6f} | {sign:>5}")

        if abs(f_m) < tol:

            return x_m

        if product < 0.0:
            x_u = x_m
        else:
            x_l = x_m
            f_l = f_m

    print("Reached max_iter without meeting the tolerance")
    return x_m

root = bisection(1, 0)
print(f"Root ≈ {root:.6f}")
```

```python
import math

def function(x):
    return math.cos(x) - x

def derivative(x):
    return -math.sin(x) - 1

def newton_raphson(x_0, tol=1e-6, max_iter=100):
    x_i = x_0
    n = 0

    print(f"{'Iter':<5} | {'x_i':>12} | {'x_i+1':>12} | {'% Diff':>12}")
    print("-" * 50)

    while n < max_iter:
        f_i = function(x_i)
        df_i = derivative(x_i)

        if df_i == 0.0:
            print("Derivative is zero, cannot continue.")
            return None

        x_new = x_i - f_i / df_i
        n += 1

        if x_new != 0:
            percent_difference = abs((x_new - x_i) / x_new) * 100
        else:
            percent_difference = 0.0


        print(f"{n:<5} | {x_i:>12.6f} | {x_new:>12.6f} | {percent_difference:>12.6f}")

        if percent_difference < tol:
            return x_new

        x_i = x_new

    print("Warning: Max iterations reached without convergence.")
    return x_new

root = newton_raphson(0.5)
print(f"Root ≈ {root:.6f}")
```

```python
def function(x):
    return (x**2) - 5

def secant_method(x_0, x_i, tol=1e-6, max_iter=10000):
    n = 0
    print(f"{'Iter':<5} | {'x_i':>12} | {'x_i+1':>12} | {'% Diff':>12}")
    print("-" * 50)
    while n < max_iter:
        f_i = function(x_i)
        f_0 = function(x_0)

        denominator = (f_i - f_0)
        if denominator == 0.0:
            print("Denominator is zero, division by zero error.")
            return None

        x_new = x_i - (f_i * (x_i - x_0)) / denominator
        n += 1

        if x_new != 0:
            percent_difference = abs((x_new - x_i) / x_new) * 100
        else:
            percent_difference = 0.0

        if percent_difference < tol:
            return x_new
        print(f"{n:<5} | {x_i:>12.6f} | {x_new:>12.6f} | {percent_difference:>12.6f}")

        x_0 = x_i
        x_i = x_new

    print("Warning: Max iterations reached without convergence.")
    return x_new

root = secant_method(2, 3)
print(f"Root ≈ {root:.6f}")
```
