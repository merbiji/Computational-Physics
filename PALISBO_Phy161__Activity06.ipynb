```python
import math

def function(x):
  return 6*math.sin(x) * math.exp (-0.8*x) -1
def derivative(x):
  return 6 * math.exp(-0.8 * x) * (math.cos(x) - 0.8 * math.sin(x))

def NR_method(x_0, tol = 0.01, max_iter=1000):
  x_i = x_0
  n = 0
  print("----------------Newton Ralphson Method --------------------")
  print(f"Iter    |   x_old      |   x_new    |   Percent Difference")
  print("-"*60)

  while n<max_iter:
    f1 = function(x_i)
    d_f1 = derivative(x_i)

    if d_f1 == 0:
      print("Derivative is 0 can't find roots")
      return None

    x_new = x_i - f1/ d_f1
    n += 1

    if x_new != 0:
      percent_difference = abs((x_new - x_i) / x_new) * 100
    else:
      percent_difference = 0.0

    print(f"{n:<5}   |   {x_i:>5f}   |    {x_new:>5f}|   {percent_difference:>5f}")

    if percent_difference < tol:
      print(f"Root found at x = {x_new}")
      print("\n")
      return x_new

    x_i = x_new

  print("Warning: Max iterations reached without convergence.")

def secant_method(x_0, x_1, tol=0.01, max_iter=100):
  print("----------------Secant Method --------------------")
  print(f"Iter    |   x_old      |   x_new    |   Percent Difference")
  print("-"*60)
  n = 0
  while n < max_iter:
    f0 = function(x_0)
    f1 = function(x_1)

    denominator = (f1 - f0)
    if denominator == 0.0:
      print("Denominator is 0 can't find roots")
      return None

    x_new = x_1 - f1 * (x_1 - x_0) / denominator
    n += 1

    if x_new != 0:
      percent_difference = abs((x_new - x_1) / x_new) * 100
    else:
      percent_difference = 0.0

    print(f"{n:<5}   |   {x_1:>5f}   |    {x_new:>5f}|   {percent_difference:>5f}")

    if percent_difference < tol:
      print(f"Root found at x = {x_new}")
      print("\n")
      return x_new

    x_0 = x_1
    x_1 = x_new

NR_method(x_0=0.25)
secant_method(x_0=0.4, x_1=0.3)
```
