SINE

```python
import math

def compute_sine(x, tol = 1e-8):
  term = x
  total_sum = x
  n = 2

  while True:

        term = -term * (x**2) / ((2*n - 1) * (2*n - 2))
        total_sum += term

        if abs(term / total_sum) < tol:
            break
        n += 1

  return total_sum, n

x_vals = [0.1, 1.0, 2.0]

print("Function\tx\timax\tsum\t\t|sum-f(x)|/f(x)")
print("-" * 65)

for x in x_vals:

    sin_approx, sin_iters = compute_sine(x)
    sin_exact = math.sin(x)
    sin_err = abs(sin_approx - sin_exact) / abs(sin_exact)

    print(f"Sine\t\t{x}\t{sin_iters}\t{sin_approx:.8f}\t{sin_err:.2e}")
```

COSINE

```python
import math

def compute_cosine(x, tol=1e-8):
    term = 1.0
    total_sum = 1.0
    n = 2

    while True:
        term = -term * (x**2) / ((2*n - 2) * (2*n - 3))
        total_sum += term

        if abs(term / total_sum) < tol:
            break
        n += 1

    return total_sum, n

x_vals = [0.1, 1.0, 2.0]

print("Function\tx\timax\tsum\t\t|sum-f(x)|/f(x)")
print("-" * 65)

for x in x_vals:
    cos_approx, cos_iters = compute_cosine(x)
    cos_exact = math.cos(x)
    cos_err = abs(cos_approx - cos_exact) / abs(cos_exact)

    print(f"Cosine\t\t{x}\t{cos_iters}\t{cos_approx:.8f}\t{cos_err:.2e}")
```
